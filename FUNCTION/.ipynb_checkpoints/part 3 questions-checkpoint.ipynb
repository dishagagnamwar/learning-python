```python
#14
def reverse():
    s=input("enter a string")
    return s[::-1]

result=reverse()
print(result)
```

```python
#15check palindrome 
def palindrome(n):
    return n.lower()==n.lower()[::-1]
print(palindrome("mom"))
```

```python
#16 count vowels
def vowel():
    vowel="aeiouAEIOU"
    s=input("enter a string")
    count=0
    for i in s:
        if i in vowel:
            count+=1
    else:
        return count

print(vowel())
```

```python
#17
def largest():
    a=[10,20,30,40,50,60]
    largest =a[0]
    for i in a:
        if i>largest:
            largest=i
            
    return largest
        
print(largest())
```
