```python
# there are 4 TYPES OF NORMAL FUNCTION
```

```python
#1. no arguments no return type

#in this type we not pass arguments with functions 
#also not return anything 

def greet():       #creating  function
    print("hello")

greet()           #calling function
```

```python
def greet():       #creating  function
    print("hello")

greet()    

a=greet()
print(a)
```

```python
#2. no argument with return type 

def get_pi():
    return 3.14

result=get_pi()
print(result)
```

```python
#2. no argument with return type 

def xyz():
    a=10
    return a
result=xyz()
print(result)
```

```python
#3. argument without return type 
def addition(a,b,c):       #we pass arguments
    result=a+b+c
    print(result)

addition(10,20,30)
```

```python
#3. argument without return type 
def addition(a,b,c):       #we pass arguments
    result=a+b+c
    print(result)

x=int(input("enter number"))
y=int(input("enter number"))
z=int(input("enter number"))
addition(x,y,z)
```

```python
#4. argument with return type 
def addition(a,b,c):       #we pass arguments
    result=a+b+c
    return result

x=int(input("enter number"))
y=int(input("enter number"))
z=int(input("enter number"))
addition(x,y,z)
```

```python
#4. argument with return type 
def addition(a,b,c):      
    return a+b+c
    
x=int(input("enter number"))
y=int(input("enter number"))
z=int(input("enter number"))
result=addition(x,y,z)
print(result)
```

```python
#without arguments with return 

def addition ():
    a=10
    b=20
    c=30
    return a+b+c

result =addition()
print(result)
```

```python
def addition ():
    a=int(input("enter number"))
    b=int(input("enter number"))
    c=int(input("enter number"))
    return a+b+c

result =addition()
print(result)
```

```python
def multiply():
    a=int(input("enter number"))
    b=int(input("enter number"))
    c=int(input("enter number"))
    return a*b*c

result =multiply()
print(result)
```

```python
def compare():
    a=int(input("enter number"))
    b=int(input("enter number"))
    if a>b:
        return a 
    else:
        return b 
    
result =compare()
print(result)
```

```python
def compare():
    a=int(input("enter number"))
    b=int(input("enter number"))
    c=int(input("enter number"))
    if a>b and a>c:
        return a 
    elif b>a and b>c:
        return b
    else:
        return c
    
result =compare()
print(result)
```

```python
def compare(a,b):
    if a>b:
        return a
    else:
        return b 

result=compare(10,compare(20,30))
print(result)
```
