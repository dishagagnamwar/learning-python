```python
#6 calculate factorial 
def factorial(n):
    if n<0:
        return "invalid"
    result=1
    for i in range (1,n+1):
        result*=1
    return result

result=factorial(5)
print(result)
```

```python
#7
def prime():
    n=int(input("enter number to check "))
    is_prime=True 
    if n<=1:
        return "not a prime"
    else:
        for i in range (2,n):
            if n%i==0:
                is_prime=False
                break 
if is_prime: 
    return "prime"
else:
    return "not prime"

result=prime()
print(result)
```

```python
#9 fabonacci series 
def fabo(n):
    if n<=0:
        return "invalid"
    else:
        a=0
        b=1
    for i in range(n):
        print(a, end=" ")
        c=a+b
        a=b
        b=c
result=fabo(10)
print(result)
```

```python
#16 count vowels
def vowel():
    vowel="aeiouAEIOU"
    s=input("enter a string")
    count=0
    for i in s:
        if i in vowel:
            count+=1
    else:
        return count

print(vowel())
```

```python
#11 sum of digits 
# n=12345%10=5     sum=sum+5   sum=5   n=12345//10=1234
# n=1234%10=5     sum=5+4     sum=9    n=12345//10=123
# n=123%10=5     sum=4+9     sum=12   n=12345//10=12
# n=12%10=5     sum=9+12   sum=14   n=12345//10=1
# n=1%10=5     sum=12+14   sum=15   n=12345//10=0

def sum_num(n):
    if n==0:
        return 0
    else:
        s=0
        temp=n
        while temp>0:
            digit=n%10
            s=s+digit 
            n=n//10
    return s

print(sum_num(12345))
```
