```python
#1 filter even numbers
n=[10,15,22,31,40,51,60]
even=list(filter(lambda n:n%2==0,n))
print(even)
```

```python
#2 filter odd numbers 
n=[11,20,33,44,55,68,71]
odd=list(filter(lambda n:n%2!=0,n))
print(odd)
```

```python
#3 number greater than 50 
n=[20,55,32,80,45,90,12]
great=list(filter(lambda n:n>50,n))
print(great)
```

```python
#4 number less than 30 
n=[10,45,25,60,18,35,5]
less=list(filter(lambda n:n<30,n))
print(less)
```

```python
#5 positive numbers 
n=[-10,20,-5,30,0,-15,40]
positive=list(filter(lambda n:n>0,n))
print(positive)
```

```python
#6 negative numbers 
n=[-10, 20, -5, 30, 0, -15, 40]
negative=list(filter(lambda n:n<0,n))
print(negative)
```

```python
#7 number divisible by 5 
n=[10, 13, 25, 32, 40, 44, 55, 61]
divisible=list(filter(lambda n:n%5==0,n))
print(divisible)
```

```python
#8 divisible by 3 
n= [10, 12, 17, 21, 25, 30, 41]
divisible=list(filter(lambda n:n%3==0,n))
print(divisible)
```

```python
#9 names starting with "A"
names = ["Amit", "Rahul", "Ankit", "Priya", "Ajay", "Neha"]
a=list(filter(lambda names:names.startswith("A"),names))
print(a)
```

```python
#10 names with more than 5 characters 
names = ["Amit", "Rahul", "Alexander", "Priya", "Gagan", "Rohit"]
char=list(filter(lambda names:len(names)>5,names))
print(char)
```
