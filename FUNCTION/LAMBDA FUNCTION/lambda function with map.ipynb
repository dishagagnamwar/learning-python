```python
#lambda function with map ()
#map ()- when we want to apply function to every element of iterables (list,tuple ,set )
#then we use map()

#syntax- map(function,iterables)

a=[2,4,6,8]
square=map(lambda x:x**2,a)
print(list(square))
```

```python
#map()- transforms every element of iterable 
#filter()- it is used to filter some elements from iterables based on some condition
#syntax- filter(function , iterables)

a=[1,2,3,4,5,6,7,8]
even=filter(lambda x:x%2==0,a)
print(list(even))
```

```python
a=[1,2,3,4,5,6,7,8]
odd=filter(lambda x:x%2!=0,a)
print(list(odd))
```

```python
#divisible by 3 
a=[1,2,3,4,5,6,7,8]
three=filter(lambda x:x%3==0,a)
print(list(three))
```

```python
name=["riya","gagan","ruchi","bhavya","aman"]
words_morethan_5=filter(lambda x:len(x)==5,name)
print(list(words_morethan_5))
```

```python
name=["amit","rahul","ankit","priya","ajay","neha"]
result=filter(lambda name:name.startswith("a"),name)
print(list(result))
```

```python
name=["riya","gagan","ruchi","bhavya","aman"]
upper=map(lambda x:x.upper(),name)
print(list(upper))
```
