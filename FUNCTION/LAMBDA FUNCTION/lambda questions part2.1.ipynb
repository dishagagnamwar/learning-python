```python
#11 ends with "a"
names = ["Riya", "Amit", "Pooja", "Rahul", "Neha", "Gagan"]
end=list(filter(lambda names:names.endswith("a"),names))
print(end)
```

```python
#12 words with exactly 4 characters 
words = ["book", "pen", "tree", "apple", "code", "cat"]
char=list(filter(lambda words:len(words)==4 ,words))
print(char)
```

```python
#13 filter non empty string 
words = ["Python", "", "Django", "", "React", "SQL"]
char=list(filter(lambda words:words!="",words))
print(char)
```

```python
#14 filter passing marks
marks = [35, 67, 22, 89, 40, 38, 76]
passing=list(filter(lambda marks:marks>=40,marks))
print(passing)
```

```python
#15 marks above 75
marks = [55, 78, 92, 65, 88, 45, 76]
passing=list(filter(lambda marks:marks>75,marks))
print(passing)
```

```python
#16 price below 500
prices = [250, 750, 450, 1200, 350, 900, 150]
below500=list(filter(lambda prices:prices<500,prices))
print(below500)
```

```python
#17 salaries above 50,000
salaries = [35000, 55000, 48000, 75000, 62000, 42000]
above=list(filter(lambda salaries:salaries>50000,salaries))
print(above)
```

```python
#18 filter adult ages 
ages = [12, 18, 25, 15, 32, 17, 21]
adult=list(filter(lambda ages:ages>18,ages))
print(adult)
```

```python
#19 filter long words 
words = ["Python", "Django", "Programming", "SQL", "JavaScript", "AI"]
long=list(filter(lambda words:len(words)>6,words))
print(long)
```
