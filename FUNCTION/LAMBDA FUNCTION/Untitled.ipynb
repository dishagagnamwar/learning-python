```python
#11 meter to centimeter
m=[1,2,5,10]
centimeter=map(lambda m:m*100,m)
print(list(centimeter))
```

```python
#12 add gst of 18%
price=[100,200,500,1000]
final_price=map(lambda price:price+(price*18/100),price)
print(list(final_price))
```

```python
#13 convert names to uppercase 
name=["gagan","rahul","amit","priya"]
upper=map(lambda name:name.upper(),name)
print(list(upper))
```

```python
#14 convert names to lowercase
name=['GAGAN', 'RAHUL', 'AMIT', 'PRIYA']
lower=map(lambda name:name.lower(),name)
print(list(lower))
```

```python
#15 find lenght of every word 
word=["python","django","sql","react"]
length=map(lambda word:len(word),word)
print(list(length))
```

```python
#16 add prefix to every name 
name=["rahul","amit","gagan"]
prefix=map(lambda name:"mr."+name ,name)
print(list(prefix))
```

```python
#17 add prefix to every city 
city=["bhopal","indore","jaipur","delhi"]
prefix=map(lambda city:"City:"+city ,city)
print(list(prefix))
```

```python
#18 calculate simple interest for each record 
principal=[1000,2000,5000,10000]
rate=5
time=2
SI=map(lambda principal:(principal*rate*time)/100,principal)
print(list(SI))
```

```python
#19 calculate area of square 
side=[2,4,5,10]
area=map(lambda side:side*side ,side)
print(list(area))
```
