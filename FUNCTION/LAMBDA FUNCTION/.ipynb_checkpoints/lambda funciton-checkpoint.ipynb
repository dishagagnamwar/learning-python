```python
#lambda function
#it is an anonymous function , function is used without name , 
#syantax lambda arguments ; expression 
#it is used for small tasks
#we write code in single line by using lambda function 

def square(n):
    return n*n

print(square(5))

#square  of number using lambda function 
#example of lambda function with single argument 

square= lambda x:x*x      #lambda argument : expression
print(square(4))
```

```python
#lambda function with multiple arguments
#sum of two numbers

result=lambda a,b:a+b
print(result(10,20))
```

```python
#substraction of two numbers

result=lambda a,b:a-b
print(result(10,20))
```

```python
#multiple of two numbers

result=lambda a,b:a*b
print(result(10,20))
```

```python
#sum of two numbers

result=lambda a,b,c:a+b+c
print(result(10,20,2))
```

```python
#convert string to upper case 
upper=lambda s:s.upper()
print(upper("gagan"))
```
