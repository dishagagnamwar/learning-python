```python
#1 double every number
a=[2,4,6,8]
double=map(lambda x:x*x,a)
print(list(double))
```

```python
#2 add 10 to every number 
a=[5,10,15,20]
add_10=map(lambda a:a+10,a)
print(list(add_10))
```

```python
#3 substract 5 to every number 
a=[5,10,15,20]
sub_5=map(lambda a:a-5,a)
print(list(sub_5))
```

```python
#4 multiply every number by 3
a=[2,4,6,8]
mul_3=map(lambda a:a*3,a)
print(list(mul_3))
```

```python
#5 square of every number 
a=[1,2,3,4,5]
square=map(lambda a:a*a,a)
print(list(square))
```

```python
#6 cube of every number 
a=[1,2,3,4,5]
cube=map(lambda a:a*a,a)
print(list(cube))
```

```python
#7 number to string 
a=[10,20,30,40]
string=map(lambda a:str(a),a)
print(list(string))
```

```python
#8 remainder find after dividing by5
n=[10,12,17,23,30]
divide=map(lambda n:n%5 ,n)
print(list(divide))
```
