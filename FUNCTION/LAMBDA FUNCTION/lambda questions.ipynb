```python
#1 add two numbers 
result=lambda a,b:a+b
print(result(10,20))
```

```python
#2 substract two numbers
result=lambda a,b:a-b
print(result(10,20))
```

```python
#3 multiply two numbers
result=lambda a,b:a*b
print(result(10,20))
```

```python
#4 divide two numbers 
result=lambda a,b:a//b
print(result(10,2))
```

```python
#5 square of a number 
square= lambda x:x*x     
print(square(4))
```

```python
#6 cube of a number 
square= lambda x:x*x*x    
print(square(4))
```

```python
#7 finnd the remainder 
remainder =lambda a,b:a//b
print(remainder(10,3))
```

```python
#8 finnd the quotient
quotient =lambda a,b:a%b
print(quotient(10,3))
```

```python
#9 add 10 to a number 
number=lambda a:a+10
print(number(5))
```

```python
#10 convert c to f 
temp=lambda c:(c*9/5)+32
print(temp(4))
```

```python
#11 calculate SI
si=lambda p,r,t:(p*r*t)/100
print(si(2000,6,2))
```

```python
#12 area of rectangle 
area=lambda l,b:l*b
print(area(20,6))
```

```python
#13 perimeter of rectangle 
area=lambda l,b:2*(l+b)
print(area(20,6))
```

```python
#14 join first name and last name
firstname=input("firstname")
lastname=input("lastname")
fullname=lambda firstname, lastname:(firstname +" "+ lastname)
print(fullname(firstname,lastname))
```

```python
#15 calculate total price 
price=int(input("enter price"))
quantity=int(input("enter quantity"))
total_price=lambda price,quantity:(price*quantity)
print(total_price(price,quantity))
```
