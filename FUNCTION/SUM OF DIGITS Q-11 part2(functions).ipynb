```python
#11 sum of digits 
# n=12345%10= 5 sum=sum+5   sum=5   n=12345//10=1234

def sum_num(n):
    if n==0:
        return 0
    else:
        s=0
        temp=n
        while temp>0:
            digit=n%10
            s=s+digit 
            n=n//10
    return s

print(sum_num(12345))
```
