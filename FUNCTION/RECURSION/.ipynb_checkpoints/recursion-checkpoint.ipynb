```python
#print natural numbers 

def print_no(n):
    if n<=1:
        return 1       #base condition / stop condition
    print(n)
    return print_no(n-1)

print(print_no(10))
```

```python
def print_no(n):
    if n<=1:
        return 1   #base condition / stop condition
    print_no(n-1)
    print(n)
    
print(print_no(10))
```

```python
#sum of n natural numbers using recursion 
def sum_num(n):
    if n==0:
        return 0
    return n+sum_num(n-1)

print(sum_num(10))
    
```

```python
# multiplication / factorial 
def factorial(n):
    if n==0:
        return 1
    return n**factorial(n-1)

print(factorial(10))
```

```python
def fabonacci(n):
    if n<=1:
        return n
    return fabonacci(n-1)+fabonacci(n-2)

print(fabonacci(5))
```

```python
def square(a):
    result=[]
    for i in a:
        result.append(i**2)
    return result
print(square([2,4,6,8]))
```
