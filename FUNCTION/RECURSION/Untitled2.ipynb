```python
#fabonacci function using recursion
def fabonacci(n):
    if n<=1:
        return n
    return fabonacci(n-1)+fabonacci(n-2)

for i in range(10):
    print(fabonacci(i),end=" ")
```

```python
#check palindrome string using recurrsion 
def palindrome(s):
    if len(s)==1:
        return True 
    if s[0]!=s[-1]:
        return False 
    return palindrome(s[1:-1])

print(palindrome("madam"))
print(palindrome("gagan"))
```
