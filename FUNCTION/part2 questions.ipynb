```python
#6 calculate factorial 
def factorial(n):
    if n<0:
        return "invalid"
    result=1
    for i in range (1,n+1):
        result*=1
    return result

result=factorial(5)
print(result)
```

```python
##7
def prime():
    n=int(input("enter number to check "))
    is_prime=True 
    if n<=1:
        return "not a prime"
    else:
        for i in range (2,n):
            if n%i==0:
                is_prime=False
                break 
if is_prime: 
    return "prime"
else:
    return "not prime"

result=prime()
print(result)
```

```python
#9 fabonacci series 
def fabo(n):
    if n<=0:
        return "invalid"
    else:
        a=0
        b=1
    for i in range(n):
        print(a, end=" ")
        c=a+b
        a=b
        b=c
result=fabo(10)
print(result)
```

```python
#13
def palindrome(n):
    if n<=0:
        return "not palindrome"
    else:
        temp=n
        s=0
        while temp>0:
            digit=temp%10
            s=s*10+digit
            temp=temp//10

        if s==n:
            return "palindrome"
        else:
            return "not palindrome"

print(palindrome(12321))
```
