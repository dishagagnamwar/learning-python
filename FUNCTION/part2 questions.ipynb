```python
#6 calculate factorial 
def factorial(n):
    if n<0:
        return "invalid"
    result=1
    for i in range (1,n+1):
        result*=1
    return result

result=factorial(5)
print(result)
```

```python
#9 fabonacci series 
def fabo(n):
    if n<=0:
        return "invalid"
    else:
        a=0
        b=1
    for i in range(n):
        print(a, end=" ")
        c=a+b
        a=b
        b=c
result=fabo(10)
print(result)
```

```python
#10 reverse a number
def rev():
    n=input("enter a number")
    return n[::-1]

result=rev()
print(result)
    
```
