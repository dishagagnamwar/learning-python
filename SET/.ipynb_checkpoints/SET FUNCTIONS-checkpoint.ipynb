```python
#1.  len(set) Definition: Returns the total number of elements in a set.
#Syntax: len(set_name)
s = {10, 20, 30} 
print(len(s)) 
        
```

```python
#2.  max(set) Definition: Returns the largest element in a set.
#Syntax: max(set_name)
s = {10, 40, 20} 
print(max(s)) 
         
```

```python
#3.  min(set) Definition: Returns the smallest element in a set.
#Syntax: min(set_name)
s = {10, 40, 20} 
print(min(s)) 
```

```python
#4 sum(set) Definition: Returns the sum of all numeric elements.
#Syntax: sum(set_name)
s = {10, 20, 30} 
print(sum(s)) 
```

```python
#5 sorted(set) Definition: Returns a sorted LIST from the set.
#Syntax: sorted(set_name)
s = {30, 10, 20} 
print(sorted(s))
```

```python
#6 set() Definition: Creates a new set or converts an iterable into a set.
#syntax: set(iterable)
print(set([1,2,2,3])) 
```
