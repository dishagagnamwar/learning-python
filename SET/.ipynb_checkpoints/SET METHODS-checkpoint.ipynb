```python
#methods of set 
#1 ADD
#adding element -add() - it is used to add element 
#update - it is used to multiple values add 

a={1,2,3,4}
a.add(10) # add single element 
print(a)
```

```python
#2 UPDATE
a.update({100,200}) # add multiple elements
print(a)
```

```python
#3 remove() Definition: Removes the specified element. Raises KeyError
#if the element is not present.
#Syntax: set_name.remove(value)
a={1,2,3,4}
a.remove(3)
print(a)
```

```python
#4 discard() Definition: Removes the specified element. Does NOT raise
#an error if the element is absent.
#Syntax: set_name.discard(value)
a={1,2,3,4}
a.discard(5)
print(a)
```

```python
#5.  pop() Definition: Removes and returns a random element.
#Syntax: set_name.pop()
a={1,2,3,4}
a.pop()
print(a)
```

```python
#6 clear() Definition: Removes all elements from the set.
#Syntax: set_name.clear()
a={1,2,3,4,5,6}
a.clear()
print(a)
```

```python
#7 copy() Definition: Returns a shallow copy of the set.
#Syntax: new_set = set_name.copy()
a={1,2,3,4,5,6}
b=a.copy()
print(b)
```

```python
#8.  union() Definition: Returns a new set containing all unique elements.
#Syntax: set1.union(set2) or set1 | set2
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.union(s2)
print(s)
```

```python
#9.  intersection() Definition: Returns common elements.
#Syntax: set1.intersection(set2) or set1 & set2
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s=s1.intersection(s2)
print(s)
```

```python
#10. difference() Definition: Returns elements present only in the first set.
#Syntax: set1.difference(set2) or set1 - set2
s1={1,2,3,4,}
s2={4,5,6,7,8}
s=s1.difference(s2)
print(s)
```

```python
#11. symmetric_difference() Definition: Returns elements that are present
#in either set but not both.
#Syntax: set1.symmetric_difference(set2) or set1 ^ set2
s1={1,2,3,4,}
s2={4,5,6,7,8}
s=s1.symmetric_difference(s2)
print(s)    #did not gave 4
```

```python
#12 isdisjoint() Definition: Returns True if both sets have no common elements.
#Syntax: set1.isdisjoint(set2)
s1={1,2,3,4,}
s2={4,5,6,7,8}
s=s1.isdisjoint(s2)
print(s) 
```

```python
#13 issubset() Definition: Returns True if every element of the first
# set exists in the second.
#Syntax: set1.issubset(set2)
s1={1,2,3,4,}
s2={4,5,6,7,8}
s=s1.issubset(s2)
print(s) 
```

```python
#14 issuperset() Definition: Returns True if the first set contains all
# elements of the second.
#Syntax: set1.issuperset(set2)
s1={1,2,3,4,5,6,7,8}
s2={4,5,6,7,8}
s=s1.issuperset(s2)
print(s) 
```

```python
#15 del Definition: delete set from memory
#Syntax: del set_name
a={1,2,3,4,5,6}
del a 
print(a)
```
