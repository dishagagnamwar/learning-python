```python
#10
a=[10,20,10,30,20,40,10]
unique_list=[]
for item in a:
    if item not in unique_list:
        unique_list.append(item)
duplicates_removed=len(a)-len(unique_list)

print(f"origina list ={a}")
print(f"list without duplicates = {unique_list}")
print(f"duplicate elements removed = {duplicates_removed}")
```
