```python
#control flow statement
#control flow decides in which order python executes in statement . 
#2 types 1.conditional 2. looping

#conditional : it works based on condition (if , if else , if elif else , nested if)

#IF : only works for TRUE ondition 
age=int(input("enter age :"))

if age>=18:
    print("eligible for vote")
    
```

```python
#IF ELSE : works with false condition 
age=int(input("enter age :"))

if age>=18:
    print("eligible for vote")

else:
    print("not eligible for vote")
```

```python
#IF ELIF ELSE : if there are multiple true condition

marks=int(input("enter marks:"))
if marks>=85:
    print("A grade")
elif marks>=70 and marks<85:
    print("B grade")
elif marks>=50 and marks<70:
    print("C grade")
elif marks>=35 and marks<50:
    print("D grade")
elif marks<0:
    print("enter valid marks")
else:
    print("fail")
```

```python
# NESTED IF : if block inside if 

age=int(input("enter age :"))
citizenship=input("enter citizenship:")
if age>=18:
    if citizenship=="indian":
        print("you are eligible for vote")
    else:
          print("indian citizenship required for vote")
else:
    print("age should be greater than 18")


```

```python
#same condition with AND OPERATOR 
age=int(input("enter age :"))
citizenship=input("enter citizenship:")

if age>=18 and citizenship=="indian":
    print("eligible")
else:
    print("not eligible")

```

```python
#SHORTHAND IF 
age=int(input("enter age :"))
if age>=18:print("eligible")
```

```python
#shorthand if else 
#ternary operator
age=int(input("enter age :"))
result="eligible" if age>=18 else "not eligible"
print(result)
```
