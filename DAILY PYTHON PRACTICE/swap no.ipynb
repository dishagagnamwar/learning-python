```python
#method 1 without using a 3rd variable 
a=10
b=20
print(f"number before swapping a is {a} and b is {b}")
a=a+b
b=a-b
a=a-b
print(f"number after swapping a is {a} and b is {b}")
```

```python
#method 2 without using third variable 
a=10
b=20
print(f"number before swapping a is {a} and b is {b}")
a,b=b,a
print(f"number after swapping a is {a} and b is {b}")
```

```python
#method 3 with using third variable 
a=10
b=20
print(f"number before swapping a is {a} and b is {b}")
c=a
a=b
b=c
print(f"number after swapping a is {a} and b is {b}")
```
