```python
year=int(input("enter year"))
if year%4==0:
    if year%100==0:
        if year%400!=0:
            print("leap year")
        else:
            print("not leap year")
    else:
        print("leap year")
else:
    print("not leap year")
```

```python
#method 2
year=int(input("enter year"))
if year%4==0 or year%100==0 and year%100!=0:
    print("leap year")
else:
    print("not leap year")
```
