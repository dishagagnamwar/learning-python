```python
#3
password=input("enter password")
is_upper=False      #helper variable or signal
is_digit=False       #helper variable or signal
for i in password:    #direct looping 
    if i.isdigit():
        is_digit=True
    elif i.isupper():
        is_upper=True

if len(password)>8 and is_digit and is_upper:
    print("valid password")
else:
    print("invalid password")
```

```python
#4
s=input("write a sentence")
s=s.lower()
vowel_count=0
consonent_count=0
for i in s:
    if i in 'aeiou':
        vowel_count+=1
    elif i not in 'aeiou' and i!=" ":
        consonent_count+=1
print(vowel_count)
print(consonent_count)
```
