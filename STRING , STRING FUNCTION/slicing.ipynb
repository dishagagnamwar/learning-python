```python
#slicing 
#string(start:end:step)
s="decodeData"
print(s[2:])
print(s[2:6:])
print(s[:6:])
print(s[::3]) #every 3
print(s[1:8:2])
print(s[-1::])
print(s[::-1]) #reverse string
```

```python
s="DataAnalyst"
print(s[2:7:])
print(s[1::2])
print(s[::-1])
```

```python
#looping on string
#direct iteration
#indirect iteration

s="DecodeData"
for i in s:          #direct iteration
    print(i,end="  ")
```

```python
for i in range(len(s)):   #indirect iteration
    print(s[i])
```

```python
#find length of string without using len()
s="decodedata"
count=0
for i in s:
    count=count+1
    print(i,count)
print(count)
```

```python
#reverse string without slicing
s="decodedata"
rev=""
for i in s:
    rev=i+rev
print(rev)
```
