```python
#loop control 
#2 types 
#1.for loop (we know the condition , 2.while loop (based onn a condition)
#FOR LOOP, 

for i in range (1,11,2):     #forward for loop
    print(i, end="")
```

```python
for a in range(11,0,-1):      #reverse loop 
    print(a, end=" ")
```

```python
#FACTORIAL OF GIVEN NO 
number=int(input("enter a number:"))
fact=1 
for i in range (1,number+1):
    fact=fact*i 
print(fact)
```

```python
number=int(input("enter a number:"))
fact1=1 
for j in range (number, 0, -1):
    fact1=fact1*j 
print(fact1)
```

```python
# question 18
a=int(input("enter side a:"))
b=int(input("enter side b:"))
c=int(input("enter side c:"))
if a==b and a==c and b==c:
    print("equilateral")
elif (a==b and a!=c) or (b==c and b!=a) or (a==c and a!=b):
    print("isosceles")
else:print("scelene")
```
