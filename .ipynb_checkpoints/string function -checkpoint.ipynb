```python
#string function 
##funnction_name(string )
#len() , max(), min() , sorted() 

s="DecodeDataAcademy"
print(len(s)) #1st method

sum=0          #2nd method (another variable)
for i in s :
    sum=sum+1
print(sum)
```

```python
s="Gagan"
print(max(s))
print(min(s))
#ord()-to find ascii value of characters
for i in s :
    print(i,ord(i))

#chr()- convert ascii value to character
print(chr(97))
```

```python
#find maximun of string without using max()
s="Gagan"
largest=s[0]
for i in range(len(s)):
    if s[i]>largest:
        largest=s[i]
print(largest)
print(max(s))
```

```python
s="Gagan"
smallest=s[0]
for i in range(len(s)):
    if s[i]<smallest:
        smallest=s[i]
print(smallest)
print(min(s))
```

```python
#sorted()- to arrange string in ascii order
s="Gagan"
print(sorted(s))
print(sorted(s,reverse=True))      #for desending order
```

```python
#string function_name()
s="GAgan sir"
print(s.upper())
print(s.lower())
print(s.capitalize())
print(s.title())
print(s.swapcase())    #change cap to small and small to cap
```

```python
#strip(), lstrip(), rstrip()
s="@gagan@"
print(s.strip("@"))
print(s.lstrip("@"))
print(s.rstrip("@"))
```

```python
#FACTORIAL
n=int(input("enter number"))
mul=1
for i in range(n,0,-1):
    mul=mul*i
print(mul)
```

```python
#replace ("old,"new") method
s="Decode ata Academy"
print(s)
a=s.replace("ata","Data")
print(a)

s1="decode"
s2=s1.replace("o","a")
print(s2)

l=["a","b","c"]
l[0]="x"
print(l)
```

```python
#find() method- if find substring then output will index position
#if substring not found then output will be -1
s="Decode data Academy"
print(s.find("Data"))
print(s.find("data"))
```
