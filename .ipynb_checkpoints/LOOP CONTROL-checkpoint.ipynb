```python
#loop controlling methods 
#1. BREAK 
#2. CONTINUE
#3. PASS 
```

```python
#BREAK
for i in range (1,11):
    if i==5:
        break 
    print(i)
```

```python
#CONTINUE
for i in range (1,11):
    if i==5:
        continue
    print(i)
```

```python
#PASS
for i in range(1,11):
    if i==5:
        pass
    print(i)
```
