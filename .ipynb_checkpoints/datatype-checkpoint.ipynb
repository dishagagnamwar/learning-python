```python
#sequence 
#list - collection of hetrogenious data types with comma separte values 
#indexed , mutable , sequence 
a=[10, 20 , 10.5 , "gagan"]
print (a)
print(a[3]) #positive indexing 
print (a[-1]) #negative indexing 

a[2]=50
print(a)
print(a[3][2])
a[3][2]="s"
```

```python
#tuple - it is collection of element inside round bracket () with comma seperate , 
#it is indexed , immutable , sequence 
#it is executed faster than list 
#we cannot remove element from tuple after creation , but can be added 

t1=() #empty tuple 
t2=tuple() #empty tuple 
print(type(t1))
print(type(t2))

t=(10,20,10.5,"gagan")
print(t)
print(type(t))
```

```python
#mapping 
#dictionary 
#it is collection of data in key and value pair  key:value 
#unindexed , mutable , key should be unique and data should be duplicate 

d=dict() #empty dictionary 
print(d)
print(type(d))

b={"name":["yash","ruchi","disha"],"age":22}
print(b)
print (b.get("name"))
print (b.get("age"))
```

```python
#set 
#set , frozen set 
#it is a collection of unique data, heterogenious 
#it is unindexed , mutable , and perform all set operation
#it is created using {}

s=set () # empty set
print(s)
print(type(s))

m={10,20,20,20,30,"gagan"}
print(m)

#frozen set 
#frozenset is immutable 
#to remove duplicacy (removes duplicate values )
fs=frozenset (m)
print(type(fs))
```

```python
#complex number 
c=3+4j
print(type(c))
```
