```python
#11
#Write a Python program that takes an integer from the user and checks
#whether the number is even or odd.
number=int(input("enter a number"))
if number%2==0:
    print("even")
else:
    print("odd")
```

```python
#12
#Write a Python program that takes three numbers from the user and
#prints the largest number.
first=int(input("enter first number"))
second=int(input("enter second number"))
third=int(input("enter third number"))
if first>second and first>third:
    print(first)
elif second>first and second>third:
    print(second)
else:
    print(third)
    
```

```python
#13
#Write a Python program to calculate the sum of all numbers from 1 to N using a loop.
number=int(input("enter n"))
total=0
for i in range(1,number+1):
    total+=i
print(total)
```
