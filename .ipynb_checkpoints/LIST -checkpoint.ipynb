```python
#LIST - list is collection of comma seperated hetrogenious data, 
#it is indexed, mutable , ordered
#created using []

a=[] #empty list 
b=list() #empty list 
print(type(a))
print(type(b))
```

```python
c=[10,10.5,"gagan",["a","b","c"]]
print(c)
print(c[0])
print(c[1])
print(c[2])
print(c[3])
print(c[3][0])
print(c[3][1])
print(c[3][2])
```

```python
#mutable
c=[10,10.5,"gagan",["a","b","c"]]
c[1]="decode"
print(c)
```

```python
#direct looping 
c=[10,10.5,"gagan",["a","b","c"]]
for i in c:
    print(i)
```

```python
#indirect looping 
for i in range (len(c)):
    print(c[i])
```

```python
c=[10,10.5,["bhavya","aliya",["bhopal","delhi"]],["a","b","c"]]
print(c[2][2][1][2])
```

```python
c=[10,10.5,["bhavya","aliya"],["a",[ "bhopal","delhi"],"b","c"]]
print(c[3][1][0][3])
```
