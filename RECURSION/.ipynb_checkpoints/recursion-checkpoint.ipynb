```python
#print natural numbers 

def print_no(n):
    if n<=1:
        return 1       #base condition / stop condition
    print(n)
    return print_no(n-1)

print(print_no(10))
```

```python
def print_no(n):
    if n<=1:
        return 1   #base condition / stop condition
    print_no(n-1)
    print(n)
    
print(print_no(10))
```
