```python
#range function 
##range (start, stop , step)
#by default step is 0 and start is 1
print(range(5))

for i in range(5):
    print(i)

 #stop
for i in range(1,5):
    print(i)

```

```python
#step 
for i in range(1,10,2):
    print(i)
```

```python
#reverse for loop
for i in range(10,0,-2):
    print(i)
```
