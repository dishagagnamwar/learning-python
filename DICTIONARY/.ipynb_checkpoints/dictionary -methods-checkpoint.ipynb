```python
#.key()-   gives keys of dictionary 
#.values()- gives values of dictionary 
#.items()- it gives both keys and values
```

```python
students={"name":"gagan","age":32,"course":"da"}
print(students.keys())
```

```python
students={"name":"gagan","age":32,"course":"da"}
print(students.values())
```

```python
students={"name":"gagan","age":32,"course":"da"}
print(students.items())
```

```python
#LOOPING IN DICTIONARY 

#to get values 
students={"name":"gagan","age":32,"course":"da"}
for i in students.values():
    print(i,end=" ")
```

```python
#to get keys 
students={"name":"gagan","age":32,"course":"da"}
for i in students:
    print(i)
```

```python
# to get key and values
students={"name":"gagan","age":32,"course":"da"}
for i in students.items():
    print(i)
```

```python
#get() key method
# it returns the value of key 
students={"name":"gagan","age":32,"course":"da"}
print(students.get("course"))     #if not available then it returns "NONE"
print(students["course"])         #if not available then it gives ERRORE
```

```python
#adding new key value pair or element in dictionary 

#normal method
d={"name":"gagan","age":32}
print(d)
d["course"]="da" 
print(d)

#update method 
d.update({"city":"bhopal"})
print(d)

a={"city":"bhopal","qualification":"be"}
d.update(a)
print(d)
```

```python
#for changes or updates  (UPDATE METHOD)
d.update({"city":"indore"})
print(d)

#for changes or updates (NORMAL METHOD)
d["age"]=30
print(d)
```

```python
#REMOVING ELEMENTS
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}

#pop method - it is used to remove any particular key and value from dictionary 
d.pop("age")
print(d)

a=d.pop("city")
print(a)
```

```python
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}
d.popitem()       #REMOVES LAST KEY AND VALUE 
print(d)
```

```python
#DEL METHOD - it removes particular key or remove enntire dictionary 
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}
del d["city"]       #remooves particular key 
print(d)
del d             #removes entire dictionary 
print(d)
```

```python
#COPY()-to create copy of dictionary 
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}
a=d.copy()
print(a)
```

```python
#clear()- remove all element in the cell
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}
d.clear()
print(d)
print(type(d))
```

```python
#SET DEFAULT()-key and default values are passed 
d={"name":"gagan","age":32,"city":"bhopal","course":"da"}
a=d.setdefault("name","disha")
print(d)

#it updates dictionary if key is not available(adds new key)
b=d.setdefault("fees",30000)
print(d)
```

```python
#fromkeys()- iterable,default values

subject=["maths","science","english"]
d=dict.fromkeys(subject)
print(d)

#adding default values
subject=["maths","science","english"]
d=dict.fromkeys(subject,100)
print(d)
```

```python
#question - create a dict which has number as a key and value is square of that number from 1 to 10
d=dict()
for i in range(1,11):
    d[i]=i**2
print(d)
```

```python
a=[]
for i in range(1,11):
    a.append(i**2)
print(a)
```

```python
#list comprehension
result=[i**2 for i in range(1,11) if i%2==0]
print(result)
```

```python
#dictionary comprehension format 
result1={i:i**2 for i in range(1,11)}
print(result1)
```
