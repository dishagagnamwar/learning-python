```python
#all the functions work on keys 
#elements= kkey :value
e={"name":"gagan","age":32,"course":"da"}
print(len(e))       #gives total elements in dict 
print(max(e))      #if there are no integers then compare ascii value 
print(min(e))
print(sorted(e))    #it gives arranged list of keys 
print(sorted(e,reverse=True))  
```
