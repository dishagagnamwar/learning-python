```python
#11 count frequency of each character in a string 
#conncept - dictionary , loop , get()
s="decodedataacademy"
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[1]=1
print(d)
```

```python
#12 count frequency of each word in a sentence 
#concept - split () , dictionary 
s="this is the decode the data the academy"
s=s.split()
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
        
print(d)
```
