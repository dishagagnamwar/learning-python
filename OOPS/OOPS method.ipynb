```python
#method 2  of getter and sette 
#TRADITIONAL METHOD 
class account:
    def __init__(self,name):
        self.__name=name

    def get_name(self):
        print(self.__name)

    def set_name(self,newname):
        self.__name=newname

a=account("gagan")
a.get_name()
a.set_name("ruchi")
a.get_name()
```

```python
#  @property @name.setter METHOD
class account:
    def __init__(self,name):
        self.__name=name
        
    @property
    def name(self):
        print(self.__name)

    @name.setter
    def name(self,newname):
        self.__name=newname

a=account("gagan")
print(a.name)
a.name="ruchi"
print(a.name)
```
