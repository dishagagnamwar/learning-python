```python
#types of methods 
#3 types 

#instance/normal method 


#class method - we use @classmethod (decorator) to create class method and pass
#cls as argument in that method 

        
class bank:
    bank_name="this is sbi"
    def __init__(self,balance):
        self.__balance=balance

    def withdraw(self,amount):
        self.__balance-=amount
        print("current balance after withdrwal of ",amount,"is",self.__balance)
   
    def displaybalance(self):
        print("balance is",self.__balance)
    
    @classmethod 
    def displaybank(cls):
        print("bank name is ",cls.bank_name)

    def add(a,b):
        c=a+b
        print(c)


s=bank(50000)
s.displaybank()
```
