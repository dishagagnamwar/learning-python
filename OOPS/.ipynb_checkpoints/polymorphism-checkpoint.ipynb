```python
# OVER RIDING METHOD - in pyhton , if same method present in parent and child class
#and we create object of child class then method of child class
#override the method of parent class
```

```python
#METHOD OVER RIDING 

class parent :
    def display (self):
        print("parent class")

class child(parent):
    def display(self):
        print("child class")

c=child()
c.display()
```

```python
#OVER LOADING - 2 TYPES
#1.METHOD OVERLOADING 
#2.OPERATOR OVERLOADING
```

```python
#METHOD OVERLOADING - not supported in python 
def add(a,b):
    print(a+b)
add(1,2)

def add(a,b,c):
    print(a+b+c)
add(1,2,3)
```

```python
def add(*args):
    print(sum(args))

add(1,2)
add(1,2,3)
```

```python
def add(*args,**kwargs):    #keyword arguments
    print(sum(args))        #args are converted into tuple 

add(1,2)
add(1,2,3)
```

```python
#OPERATOR OVERLOADING - the operators acts diffent with diff datatypes
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary 

    def display(self):
        print("employee name is ",self.name , "and salary is",self.salary )

e1=employee("yash",50000)
e1.display()

e2=employee("disha",80000)
e2.display()

e1.salary+e2.salary
```

```python
# when the salary is private 
#special method for operator overloading 
#DUNDER METHOD
#Double underscore method

class employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary 

    def display(self):
        print("employee name is ",self.name , "and salary is",self.__salary )

    def __add__(self,other):      #DUNDER METHOD
        return self.__salary+other.__salary 

e1=employee("yash",50000)
e1.display()

e2=employee("disha",80000)
e2.display()

print(e1+e2)
```
