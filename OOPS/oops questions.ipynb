```python
#question 1 
class students:
    def __init__(self,name,age):
     self.name=name
     self.age=age

    def display(self):
        print("name is",self.name)
        print("age is",self.age)

s1=students("disha",25)
print(s1.name)
print(s1.age)
```

```python
#question 2
class rectangle :
    def __init__(self,length,width):
        self.length=length
        self.width=width

    def area(self):
        return self.length*self.width 

r=rectangle(10,5)
print("area is",r.area())
```

```python
#question 3
class employee :
    def __init__(self,empname,salary):
        self.empname=empname
        self.salary=salary

    def display(self):
        print("name:",self.empname,"salary:",self.salary)

e1=employee("disha",10000)
e2=employee("yash",20000)
e3=employee("rahul",30000)

e1.display()
e2.display()
e3.display()
```
