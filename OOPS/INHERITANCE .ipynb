```python
# INHERITANCE :- it is a property of one class to acquire properties and methods of another class
# there are 5 types 
#1.single level 
#2. multi level - grandfather , father , child 
#3. multiple - father , mother , child 
#4. heirarchial - 
#5. hybrid
```

```python
#SINGLE LEVEL
class Parent:
    def displayParent(self):
        print("this is parent class")

class child(Parent):
    def displayChild(self):
        print("this is child class")

c=child()
c.displayChild()
c.displayParent()
```

```python
# MULTI LEVEL inheritance

class A:
    def displayA(self):
        print("class a")

class B(A):
    def displayB(self):
        print("class b")

class C(B):
    def displayC(self):
        print("class c")

c=C()
c.displayC()
c.displayB()
c.displayA()
```

```python
# MULTIPLE LEVEL 

class A:
    def displayA(self):
        print("FATHER")

class B:
    def displayB(self):
        print("MOTHER")

class C(A,B):
    def displayC(self):
        print("CHILD")

c=C()
c.displayC()
c.displayB()
c.displayA()
```

```python
# HEIRARCCHIAL 

class A:
    def displayA(self):
        print("FATHER")

class B(A):
    def displayB(self):
        print("CHILD 1 ")

class C(A):
    def displayC(self):
        print("CHILD 2 ")

c=C()
b=B()
c.displayA()
b.displayA()
```

```python
# HYBRID 
class A:
    def displayA(self):
        print("PARENT")

class B(A):
    def displayB(self):
        print("CHILD OF A ")

class C(A):
    def displayC(self):
        print("CHILD OF A")

class D(B,C):
    def displayA(self):
        print("CHILD OF B AND C")

b=B()
c=C()
d=D()
b.displayA()
c.displayA()
d.displayB()
d.displayC()
d.displayA()
```

```python
# MRO - method resolution order 
#method of OVERWRITING (POLYISM)(DIAMOND PROBLEM)
class A:
    def display(self):
        print("PARENT")

class B(A):
    def display(self):
        print("CHILD OF A ")

class C(A):
    def display(self):
        print("CHILD OF A")

class D(B,C):
    pass

d=D()
d.display()
print(D.mro())
print(B.mro())
```
