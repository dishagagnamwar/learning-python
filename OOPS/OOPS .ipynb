```python
class person:
    def __init__(self,name,age): 
        self.name=name
        self.age=age

p1=person("gagan",32)
print(p1.age)
print(p1.name)
```

```python
class account:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance

account1=account("gagan",50000)
print(account1.name)
print(account1.balance)

account1.balance=-5000
print(account1.balance)
```

```python
# ENCAPSULATION - to secure data , access specifier
# wrapping data and method in single unit(class) and restrict direct access of object data outside the class
# ACCCESS SPECIFIER - 

#3 types of variable 
#public variable =  self.name=name
#protected variable =  self._name=name
#private variable  =  self.__name=name   (outside class it is not available)
```

```python
class account:
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance

    def getbalance(self):
        print("balance is",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount
        
a1=account("gagan",5000)
print(a1.name)
a1.getbalance()
a1.setbalance(20000)
a1.getbalance()
```
