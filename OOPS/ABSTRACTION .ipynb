```python
#in abstraction we hide implimentation details and show only relevant data to user 
#also used to enforce some rules to child class 

#for abstraction we use "abc" module 

#abc= this is module to achieve abstarction
#ABC=" abstract base class"-----> it provides common structure to child classes 
#abstract method = this is decorator and it is used to create method that should be 
#predefined in child class 

from abc import ABC, abstractmethod

class bank(ABC):      #abstract class
    @abstractmethod       #decorator
    def withdraw(self):
        pass 

class sbibank(bank):
    def __init__(self,balance):
        self.__balance=balance

    def withdraw(self,amount):
        self.__balance-=amount
        print("current balance after withdrwal of ",amount,"is",self.__balance)

    def displaybalance(self):
        print("balance is",self.__balance)

class ubibank(bank):
    pass                              #error here because withdraw method is not defined 
    
s=sbibank(50000)
s.displaybalance()
s.withdraw(30000)
u=ubibank()
```
