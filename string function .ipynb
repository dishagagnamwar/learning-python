```python
s="Gagan"
print(max(s))
print(min(s))
#ord()-to find ascii value of characters
for i in s :
    print(i,ord(i))

#chr()- convert ascii value to character
print(chr(97))
```

```python
#find maximun of string without using max()
s="Gagan"
largest=s[0]
for i in range(len(s)):
    if s[i]>largest:
        largest=s[i]
print(largest)
print(max(s))
```

```python
s="Gagan"
smallest=s[0]
for i in range(len(s)):
    if s[i]<smallest:
        smallest=s[i]
print(smallest)
print(min(s))
```

```python
#sorted()- to arrange string in ascii order
s="Gagan"
print(sorted(s))
print(sorted(s,reverse=True))      #for desending order
```
