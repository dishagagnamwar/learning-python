```python
#type casting 
#tuple to list 

t=(10,20,30,40)
a=list(t)
print(a)
print(type(a))
```

```python
#method of tuple 
#count
t=(10,20,30,40,20,20,50)
print(t.count(20))
```

```python
#index method - to find inndex position
t=(10,20,30,20,20,50)
print(t.index(20))


#enumarate method - it gives index value pair
t=(10,20,30,40,20,20,50)
for i in enumerate(t):
    print(i)
```

```python
#zip() method - it is used to combine multiple tuple 

names=("ruchi","bhavya","alia","laxmi")
courses=("da","ds","ml","ba")
fees=(25000,30000,35000,40000)
for i in zip(names,courses,fees):
    print(i)
```
