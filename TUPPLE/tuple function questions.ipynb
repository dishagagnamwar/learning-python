```python
#1 len(tuple) Description: Returns the total number of elements.
t=(1,2,3,4,5,6,7,8,11,23,45,54)
print(len(t))
```

```python
#2 max(tuple) Description: Returns the largest element
t=(1,2,3,4,5,6,7,8,11,23,45,54)
print(max(t))
```

```python
#3 min(tuple) Description: Returns the smallest element.
t=(1,2,3,4,5,6,7,8,11,23,45,54)
print(min(t))
```

```python
#5 sorted(tuple) Description: Returns a sorted LIST (not a tuple).
t=(1,2,3,4,5,6,7,8,11,23,45,54)
print(sorted(t))
```

```python
#4  sum(tuple) Description: Returns the sum of all numeric elements.
t=(1,2,3,4,5,6,7,8,11,23,45,54)
print(sum(t))
```

```python
#7 list(tuple) Description: Converts a tuple into a list.
t=(1,2,3,4,5,6,7,8,11,23,45,54)
a=list(t)
print(a)
```
