```python
#dictionary - data is stored in key and value pair
#it is mutable , ordered , and unindexed 
#key should be unique (intiger, string , tuple)
#value should be duplicate 

d=dict()    #first method 
e={"name":"Gagan","age":32}  #second method
print(type(d))
print(type(e))
```

```python
#acess values of dictionary 
e={"name":"Gagan","age":32} 
print(e["name"])
print(e["age"])
```

```python
#adding new key value pair in dict 
e["course"]="da"
print(e)
```

```python
#update value of key 
e["age"]=30
print(e)
```
