```python
t=(10,20,30,40)
count=0
for i in t:
    count+=1
print(count)

#with len function 
t=(10,20,30,40)
print(len(t))
```

```python
#without max function
t=(10,20,30,40)
largest=t[0]
for i in t:
    if i>largest:
        largest=1
print(largest)

#with max function
t=(10,20,30,40)
print(max(t))
```

```python
t=(10,20,30,40)
smallest=t[0]
for i in t:
    if i<smallest:
        smallest=1
print(smallest)
```

```python
#sum function 
t=(10,20,30,40)
print(sum(t))
```

```python
#sorted function 
t=(20,30,10,40)
print(sorted(t))

#reverse 
print(sorted(t,reverse=True))
```
