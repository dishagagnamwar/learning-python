```python
#11 print only even numbers from the tuple
a=(12,15,10,14,87,23,2)
for i in a:
    if i%2==0:
        print(i)
```

```python
#12
a=(12,15,10,14,87,23,2)
even=0
odd=0
for i in a:
    if i%2==0:
        even+=1
    else:
        odd+=1
print(f"even are {even}")
print(f"odd are {odd}")
```
