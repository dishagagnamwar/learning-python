```python
#1 count(value) Description: Returns the number of occurrences of a value.
t=(1,2,3,5,1,3,1,4,1,0)
print(t.count(1))
```

```python
#2 index(value) Description: Returns the index of the first occurrence.
t=(1,2,3,5,1,3,1,4,1,0)
print(t.index(2))
```

```python
#3 enumerate(tuple) Description: Returns index-value pairs.
t=(1,2,3,5,1,3,1,4,1,0)
for i in enumerate(t):
    print(i)
```
