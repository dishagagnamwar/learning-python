```python
#looping through tuple 
##direct looping 
##indirect looping (using range)



#DIRECT LOOPING 
t=(10,"gagan",10.5,[100,200],55,65)
for i in t:
    print(i,end="  ")
print()


#INDIRECT LOOPING 
for i in range(len(t)):
    print(t[i],end="  ")
```

```python
#tuple concatation 
t1=(10,20,20)
t2=(100,200,300)
t=t1+t2
print(t)
```

```python
#tuple packing - it means pack multiple data in tuple 
t=10,10.5,"gagan"
print(type(t))
print(t)
```

```python
#tuple unpacking 
t="gagan",32,"DA"
name , age , course=t
print(name)
print(age)
print(course)
```
