```python
#LIST METHODS

#append() - to insert element in list 
l=[10,20,30]
print(l)
l.append(40)
print(l)
```

```python
#extend method()- used to add multiple elements
l=[10,20,30]
l.extend((60,50))
print(l)
```

```python
#adding 2 lists
a1=[10,20,30]
a2=[50,60]
a1.extend(a2)
print(a1)
```

```python
#adding 2 lists (2nd method)
a1=[10,20,30]
a2=[50,60]
print(a1+a2)
```

```python
#insert ()-when we want to add any element at any specific index position 
#syntax - my_list.insert(index_position, element)
b=[10,20,30]
print("original",b)
b.insert(1,22)
print("modified",b)
```

```python
#Remove 
#pop()- to remove element from the list 
#syntax- pop(index_position)
#if there is not given any index position then it will remove the last element of the list 
#pop remove and return element
c=[10,20,30,40,50,60]
r=c.pop(2)
print(c)
print(r)

c.pop()
print(c)
```

```python
#remove()- it is used to remove particular element from list 
#syntax= my_list.remove(element)
#it will remove first occurance of element

d=[10,40,30,40,50,60]
d.remove(40)
print(d)
```

```python
#clear method()- to delete all element of the list 
d=[10,40,30,40,50,60]
d.clear()
print(d)
```

```python
#index method () - to find index position of any particular element 
d=[10,40,30,40,50,60]
print(d.index(40))
print(d.index(40,2))
```

```python
#count() - it is used to count the occurance of any particular element 
d=[10,40,30,40,50,60,75,60]
print(d.count(40))
print(d.count(75))
```

```python
#sort()- it is used to arrange list 
d=[10,40,30,40,50,60,75,60]
d.sort()
print(d)
d.sort(reverse=True)
print(d)
```

```python
#sorted - does not modify the list 
#sort - modify the list 
```

```python
#reverse ()- it is used to reverse list 
d=[10,40,30,40,50,60,75,60]
r=d.reverse()
print(d)
```

```python
#copy ()- to make copy of list 
d=[10,20,30,50]
e=d.copy()
print(d)
```

```python
#enumerate method()-if we want index position and value both 

a=["gagan","ruchi","alia","bhavya"]
for i in enumerate(a):
    print(i)
```

```python
#zip()- it is used to combine two or more lists

a=["gagan","ruchi","alia","bhavya","disha"]
b=["full stack","data analytics","data science","machine learning"]
for i in zip(a,b):
    print(i)
```

```python
a=[1,2,3,4,5,6]
square=[]
for i in a:
    square.append(i**2)
print(square)
```

```python
a=[1,2,3,4,5,6]
cube=[]
for i in a:
    cube.append(i**3)
print(cube)
```

```python
#LIST COMPREHENSION - it is short and effective way to create list from
#existing list in a single line 
#syntax- {expression for loop if condition }

#square of number using list comprehension
a=[1,2,3,4,5,6]
square=[i**2 for i in a]
print (square)
```

```python
b=[23,12,45,64,72,43]
even=[i for i in b if i%2==0]
print(even)
```

```python
b=[23,12,45,64,72,43]
result=["even" if i%2==0 else "odd" for i in b]
print(result)
```
