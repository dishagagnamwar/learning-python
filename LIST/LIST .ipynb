```python
c=[10,10.5,"gagan",["a","b","c"]]
print(c)
print(c[0])
print(c[1])
print(c[2])
print(c[3])
print(c[3][0])
print(c[3][1])
print(c[3][2])
```

```python
#mutable
c=[10,10.5,"gagan",["a","b","c"]]
c[1]="decode"
print(c)
```

```python
#direct looping 
c=[10,10.5,"gagan",["a","b","c"]]
for i in c:
    print(i)
```

```python
#indirect looping 
for i in range (len(c)):
    print(c[i])
```

```python
#list function and method 
#len(), max(),min(),sum(),sorted()

#LEN FUNCTION
a=[10,20,30,40,50,60]
count=0
for i in a:
    count+=1
print(len(a))
print(count)
```

```python
#MAX FUNCTION


print(max(a))

#without max function
largest=a[0]
for i in a:
    if i>largest:
        largest=i

print("largest element is :",largest)
```

```python
#MIN FUNCTION
a=[10,20,30,40,50,60]
print(min(a))

#without min function
smallest=a[0]
for i in a:
    if i<smallest:
        smallest=i

print("smallest element is :",smallest)
```

```python
#SUM FUNCTION
a=[10,20,30,40,50,60]
print(sum(a))

#without sum function
s=0
for i in a:
    s=s+i
print(s)
```

```python
#SORTED FUNCTION
a=[10,30,20,50,45,60,35]
print(sorted(a))                 #asc
print(sorted(a,reverse=True))    #desc
```
