```python
#1.LARGEST OF LIST
a=[10,20,30,40,50,60]
largest=a[0]
for i in a:
    if i>largest:
        largest=i

print("largest element is :",largest)
```

```python
#2. SMALLEST OF LIST
a=[10,20,30,40,50,60]
smallest=a[0]
for i in a:
    if i<smallest:
        smallest=i

print("smallest element is :",smallest)
```

```python
#3. SUM OF LIST
a=[10,20,30,40,50,60]
s=0
for i in a :
    s=s+i
print(s)
```

```python
#4.AVERAGE OF LIST
a=[10,20,30,40,50,60]
b=len(a)
print(b)
print(sum(a)/b)
```

```python
#5. PRINT LIST OF EVEN NUMBERS
a=[5,46,56,51,95]
even=[i for i in a if i%2==0]
print(even)
```

```python
#6.PRINT LIST OF ODD NUMBERS
a=[5,46,56,51,95]
odd=[i for i in a if i%2!=0]
print(odd)
```

```python
#7. Count Positive/Negative/Zero in list.
#Problem: Count positive, negative and zero values.
a=[10,-5,0,-7,25,0,45,-48]
for i in a:
    if i==0:
        print("zero")
    elif i<0:
        print("negative")
    else:print("positive")
```
