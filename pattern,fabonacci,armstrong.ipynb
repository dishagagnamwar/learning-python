```python
#loop cotrolling method
#1.break - if we want to stop loop at at particular condition

for i in range(1,11):
    if i==5:
        break 
    print(i)
```

```python
#continue controlling method - it skip current iterration 
for i in range(1,11):
    if i==5:
        continue 
    print(i)
```

```python
#pass- it is just a placeholder, it do nothing but it is important because 
#if we want to add some logic in future then we write pass
for i in range(1,11):
    if i==5:
      pass
    print(i)
```

```python
#patterns
for i in range(1,6): #ROW
    for j in range(1,6):# COL
        print("*",end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        if j<=i:
            print("*",end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        if j<=6-i:
            print("*",end=" ")
    print()
```

```python
for i in range (1,6):
    for j in range (1,6):
        if j>=6-i:
            print("*",end=" ")
        else:print(" ",end=" ")
    print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        if i==1 or i==5 or j==1 or j==5:
          print("*",end=" ")
        else:print(" ",end=" ")
    print()
```

```python
#swap number without using third variable (method 1)
a=10
b=20
print(f"number before swapping a is {a} and b is {b}")
a=a+b
b=a-b
a=a-b
print(f"number after swapping a is {a} and b is {b}")
```

```python
#swap number without using third variable (method 2)
a=10
b=20
print(f"number before swapping a is {a} and b is {b}")
a,b=b,a
print(f"number after swapping a is {a} and b is {b}")
```
