```python
#arithmetic operators

#addition
arr=np.array([10,2030,40])
print(arr)
arr_new=arr+4
print(arr_new)
```

```python
arr1=np.array([10,20,30,40])
arr2=np.array([1,2,3,4])
print(arr1)
print(arr2)
arr_new=arr1+arr2
print(arr_new)

arr_new1=np.add(arr1,arr2)
print(arr_new1)
```

```python
#addition in 2 dimension
arr1=np.array([[10,20,30,40],[45,86,54,53]])
arr2=np.array([[1,2,3,4],[1,8,65,8]])
print(arr1)
print(arr2)
arr_new=arr1+arr2
print(arr_new)

arr_new1=np.add(arr1,arr2)
print(arr_new1)
```

```python
#addition in 3 dimension
arr1=np.array([[[10,20,30,40],[45,86,54,53]],[[23,44,5,3],[98,21,4,56]]])
arr2=np.array([[[1,2,3,4],[1,8,65,8]],[[10,20,30,40],[45,86,54,53]]])
print(arr1)
print(arr2)
arr_new=arr1+arr2
print(arr_new)

arr_new1=np.add(arr1,arr2)
print(arr_new1)
```

```python
#substract
arr1=np.array([10,20,30,40])
arr2=np.array([1,2,3,4])
print(arr1)
print(arr2)
arr_new=arr1-arr2
print(arr_new)

arr_new1=np.subtract(arr1,arr2)
print(arr_new1)
```

```python
#reciprocal
arr=np.array([10,20,30,40])
print(arr)
arr_new=1/arr
print(arr_new)

arr_new1=np.reciprocal(arr)
print(arr_new1)
```

```python
#multiply 
arr=np.array([20,30,40,50])
print(arr)
arr_new=arr*5
print(arr_new)
```

```python
#multiply
arr1=np.array([10,60,30,80])
arr2=np.array([1,2,3,4])
print(arr1)
print(arr2)
arr_new=arr1*arr2
print(arr_new)

arr_new1=np.multiply(arr1,arr2)
print(arr_new1)
```

```python
#divide and reciprocal
arr1=np.array([50,60,45,36])
arr2=np.array([2,6,4,5])
print(arr1)
print(arr2)

arr_new=arr1/arr2
print(arr_new)

arr_new1=np.reciprocal(arr)
print(arr_new1)
```

```python
#max()
arr=np.array([100,14,67,128,97])
print(arr)
print(np.max(arr))
max_value=arr[0]
for i in arr:
    if i >max_value:
        max_value=i
print(max_value)
```

```python
arr1=np.array([[110,23,44,56],[12,34,556,67]])
print(arr1)
print(np.max(arr1))
print(np.max(arr1,axis=0))
print(np.max(arr1,axis=1))
```

```python
arr=np.array([[[10,2,34],[64,31,5]],[[68,22,36],[110,7,15]]])
print(arr)
print(np.max(arr))
print(np.max(arr,axis=0))
print(np.max(arr,axis=1))
print(np.max(arr,axis=2))
```

```python
#min
arr=np.array([100,14,67,128,97])

min_value=arr[0]
for i in arr:
    if i <min_value:
        min_value=i
print(min_value)
```

```python
arr=np.array([100,76,12,56,87,123,89])
print(arr)
print(np.max(arr),np.argmax(arr))
print(np.min(arr),np.argmin(arr))
```

```python
arr=np.array([[100,23,45,65],[78,46,67,34]])
print(arr)
print(np.max(arr),np.argmax(arr))
print(np.min(arr),np.argmin(arr))
```

```python
arr=np.array([[100,23,45,65],[78,46,67,34]])
print(arr)
print(np.argmax(arr,axis=0))
print(np.argmin(arr,axis=0))
```

```python
arr=np.array([900,10000,25,36,100])
print(arr)
print(np.sqrt(arr))
```

```python
arr=np.array([[900,10000,25,100],[81,4900,3600,400]])
print(arr)
print(np.sqrt(arr))
```

```python
arr=np.array([10,20,30,40,50])
print(arr)
print(np.cumsum(arr))
```

```python
arr=np.array([[10,20,30,40],[50,60,70,80]])
print(arr)
print(np.cumsum(arr))
```

```python
arr=np.array([[10,20,30,40],[50,60,70,80]])
print(arr)
print(np.cumsum(arr,axis=0))
print(np.cumsum(arr,axis=1))
```
