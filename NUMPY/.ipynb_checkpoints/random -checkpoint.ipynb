```python
arr=np.random.randint(10,20,(100))
print(arr)
```

```python
#copy vs view 
arr=np.array([1,2,3,4,5])
print(arr)
arr_copy=arr.copy()
print(arr_copy)
arr_copy[0]=100
print(arr_copy)
print(arr)
```

```python
arr=np.array([1,2,3,4,5])
print(arr)
arr_view=arr.view()
print(arr_view)
arr_view[2]=100
print(arr_view)
print(arr)
```

```python
#create one random integer value 1d array name age between 20 and 30 having 50 elements create boxplot also 
age=np.random.randint(20,30,(50))
print(age)
age_copy=age.copy()
print(age_copy)
```

```python
import seaborn as sns

sns.boxplot(age,color="red")
```

```python
#create one evenly spaced array having 10 elements(1d) 
arr=np.linspace(10,100,10)
print(arr)
```
