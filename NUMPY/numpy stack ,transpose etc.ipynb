```python
import numpy as np
```

```python
# stack
arr1=np.array([10,20,30,40])
arr2=np.array([100,200,300,400])
print(arr1)
print(arr2)
print("----******-----")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("----******-----")
arr_new=np.stack((arr1,arr2),axis=0)
print(arr_new)
print("----******-----")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
```

```python
# TRANSPOSE - arrayobj.T
arr=np.array([[100,200,300,400]])
print(arr)
print(arr.shape)
arr_1=arr.T
print(arr_1)
print(arr_1.shape)
```

```python
arr1=np.array([[1,2,3,4],[5,6,7,8]])
arr2=np.array([[10,11,12,13],[14,15,16,17]])
print(arr1)
print(arr2)
print("-----****------")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("-----****------")
arr_new=np.stack((arr1,arr2),axis=0)
print(arr_new)
print("-----****------")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
print("-----****------")
arr_new=np.stack((arr1,arr2),axis=2)
print(arr_new)
```

```python
# split
# np.split(arrobj,no of array)
# split works on an equal size array.
arr=np.arange(1,9)
print(arr)
l=np.split(arr,4)
for i in l:
    print(i)
```

```python
# where
arr=np.array([10,3,4,5,18,8,9,11,56])
print(arr)
print(np.where(arr%2==0))
print(np.where(arr>5))
```

```python
# np.sort(array)
arr=np.sort([100,23,45,21,46,57,12])
arr_new=np.sort(arr)
print(arr_new)
```

```python
arr=np.sort([100,23,45,21,46,57,12])
arr_new=np.sort(arr)[::-1]
print(arr_new)
```
