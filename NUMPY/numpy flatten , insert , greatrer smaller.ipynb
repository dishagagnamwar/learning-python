```python
import numpy as np
```

```python
# arrayobject.flatten() - convert multi-dimensional array into a 1D array
# it create copy of an original array.
arr=np.array([[10,20,30,40],[50,60,70,80]])
print(arr)
arr_new=arr.flatten()
print(arr_new)
arr_new[0]=100
print(arr_new)
print(arr)
```

```python
# np.insert(array,index,element)
arr=np.array([10,20,30,40,50])
print(arr)
arr1=np.insert(arr,2,100)
print(arr1)
```

```python
arr=np.array([10,20,30,40,50])
print(arr)
arr1=np.insert(arr,2,[100,200,300])
print(arr1)
```

```python
arr=np.array([[100,200,300],[400,500,600]])
print(arr)
print("-----****----")
arr_0=np.insert(arr,1,[10,20,30],axis=0)
print(arr_0)
print("-----****----")
arr_1=np.insert(arr,1,[10,20],axis=1)
print(arr_1)
```

```python
# np.delete(array,index)
arr=np.array([100,200,300,400,500])
print(arr)
arr_new=np.delete(arr,2)
print(arr_new)
arr_new1=np.delete(arr,[2,3])
print(arr_new1)
```

```python
arr=np.array([[100,200,300,400],[50,60,70,80]])
print(arr)
arr_new=np.delete(arr,1,axis=0)
print(arr_new)
arr_new1=np.delete(arr,2,axis=1)
print(arr_new1)
```

```python
arr=np.array([10,11,12,13,14,15,16,17])
result=arr[arr%2==0]
print(result)
```

```python
arr=np.array([10,11,12,13,14,15,16,17])
result=arr[arr>13]
print(result)
```

```python
arr=np.array([10,11,12,13,14,15,16,17])
result=arr[(arr>13) & (arr<16)]
print(result)
```
