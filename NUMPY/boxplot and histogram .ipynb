```python
import numpy as np
```

```python
l=[20,21,22,21,20,23,23,24,28,29,32,30,32,24,25,26,24,25,25,25,26,31,20,20,23,27,28,28,29,26]
age=np.array(l)
print(age)
```

```python
np.mean(age)
```

```python
np.median(age)
```

```python
Q1=np.quantile(age,0.25)
Q1
```

```python
Q2=np.quantile(age,0.50)
Q2
```

```python
Q3=np.quantile(age,0.75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
upper_fence=Q3+(1.5*IQR)
upper_fence
```

```python
lower_fence=Q1-(1.5*IQR)
lower_fence
```

```python
l=[]
for i in age:
    if i<lower_fence or i>upper_fence:
        l.append(i)
print(l)
```
