```python
import numpy as np
```

```python
#linspace(start,stop,[number=50... by default],endpoint=True)
#return evenly spaced array 
arr=np.linspace(0,20,5)
print(arr)
```

```python
arr=np.linspace(0,20)
print(len(arr))
```

```python
arr=np.linspace(0,20,5,endpoint=False)
print(arr)
```

```python
#zeros()- zero filled array 
arr=np.zeros(4)
print(arr)
```

```python
arr=np.zeros((2,3))
print(arr)
```

```python
#ones()- return ones filled array 
arr=np.ones(6)
print(arr)
```

```python
arr=np.ones((2,3,3))
print(arr)
```
